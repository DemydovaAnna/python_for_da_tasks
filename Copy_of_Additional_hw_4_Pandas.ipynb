10. Виведіть медіанну (медіана - це значення, менше якого є рівно 50% вибірки і більше якого є теж рівно 50% вибірки)
ЗП (Зарплата у $$$) для кожного рівня знання англійської. Які висновки можна зробити?

```python
median_salary = df.groupby('Знання англійської мови')['Зарплата у $$$ за місяць, лише ставка після сплати податків'].median().reset_index()
median_salary.columns = ['Рівень англійської', 'Медіанна ЗП ($)']
print(median_salary)
```

Спостерігається чітка позитивна кореляція: чим вищий рівень англійської, тим вища медіанна зарплата фахівця.

 Фахівці з рівнями Advanced та Upper-Intermediate заробляють суттєво більше за тих, хто має Elementary чи Pre-Intermediate, оскільки високий рівень мови дозволяє працювати безпосередньо з іноземними замовниками (США, ЄС), де ставки за годину значно вищі, а також посади високих рівнів вимагають постійної комунікації англійською, тому високий рівень мови є обов'язковою умовою для високооплачуваних позицій.

11. Є така штука як сортування значень категоріальної змінної. Нам це зараз стане в нагоді, адже рівні англійської можна посортувати.
Давайте відсотруємо рівні англ. і тайтли. Просто запустіть код нижче і проаналізуйте, що він робить.

```python
df['Знання англійської мови'] = pd.Categorical(df['Знання англійської мови'],
                      categories=['Elementary','Pre-Intermediate','Intermediate','Upper-Intermediate','Advanced'],
                      ordered=True)

df['Ваш тайтл'] = pd.Categorical(df['Ваш тайтл'],
                      categories=['Intern/Trainee', 'Junior', 'Немає тайтлу', 'Middle', 'Senior', 'Tech Lead', 'Team Lead', 'Architect', 'Head', 'Manager'],
                      ordered=True)
```

12. Зробіть зводну таблицю, де стовпці - це 'Ваш тайтл', рядки - 'Знання англійської мови', а значення - медіанна ЗП. Які висновки можна зробити?
Тут ще для зручності можна зробити градієнтне забарвлення і відформатувати значення для легшого перегляду.

```python
pivot_table = df.pivot_table(
    index='Знання англійської мови',
    columns='Ваш тайтл',
    values='Зарплата у $$$ за місяць, лише ставка після сплати податків',
    aggfunc='median',
    observed=False
)
pivot_table.style.background_gradient(cmap='YlGn').format('{:.0f}', na_rep='-')
```

Тайтл залишається головним фактором росту ЗП. Перехід від Junior до Senior дає кратне зростання доходу незалежно від мови.

Англійська мова працює як множник: на кожному тайтлі (особливо Middle та Senior) фахівці з Advanced / Upper-Intermediate мають помітно вищу медіанну ЗП, ніж їхні колеги з Elementary.

14. Як зростає зарплата з роками? Для цього нам треба відформатувати колонку "Загальний стаж роботи за спеціальністю" аналогічно до того, як ми це робили для колонки 'Знання англійської мови' і потім вивести медіанні значення для кількості років стажу.

```python
df['Загальний стаж роботи за спеціальністю'].unique()
```

```python
exp_categories = [
    'Менше як 3 місяці',
    '3 місяці',
    'Пів року',
    '1 рік',
    '1,5 року',
    '2 роки',
    '3 роки',
    '4 роки',
    '5 років',
    '6 років',
    '7 років',
    '8 років',
    '9 років',
    '10 років',
    '11 років',
    '12 років',
    '13 років',
    '14 років',
    '15 і більше років'
]
df['Загальний стаж роботи за спеціальністю'] = pd.Categorical(
    df['Загальний стаж роботи за спеціальністю'],
    categories=exp_categories,
    ordered=True )
experience_salary = df.groupby('Загальний стаж роботи за спеціальністю', observed=False)['Зарплата у $$$ за місяць, лише ставка після сплати податків'].median().reset_index()
experience_salary.columns = ['Стаж роботи', 'Медіанна ЗП ($)']
experience_salary
```

Динаміка зростання доходів від стажу:

 Найбільш динамічне зростання спостерігається в перші 5 років роботи. Медіанна ЗП зростає від *\$950* (1 рік) до *\$3500* (5 років), що відображає швидку зміну кваліфікаційних рівнів від Junior до Senior.

У діапазоні від 6 до 10 років стажу зростання сповільнюється — медіана коливається в межах *\$4000* - *\$4700*.

Після 10 років стажу медіанна ЗП стабілізується на рівні *\$4500 - $5000*. Незначні коливання в окремих роках (наприклад, 13 чи 15+ років) зумовлені меншою кількістю респондентів у цих вибірках або переходом фахівців у суміжні управлінські ролі, де структура доходу може відрізнятися.
